```python
import pandas as pd
from sklearn.model_selection import train_test_split

df_model = pd.read_csv('model_ready_house_prices.csv')

# turn SalePrice into a yes/no target: is this house priced above the median?
median_price = df_model['SalePrice'].median()
df_model['PriceAboveMedian'] = (df_model['SalePrice'] > median_price).astype(int)

print(df_model[['SalePrice', 'PriceAboveMedian']].head(10))
```
