Importing and printing data at initial

```python
import pandas as pd
df = pd.read_csv("train.csv")

print(df.shape)

# print the first 5 rows
print(df.head())
```

Checking missing values

```python
#column types and missing values
print(df.info())
```

see percent of missing values

```python

#see which columns have  how much percent missing values
missing_percent = (df.isnull().sum() / len(df)) * 100
print(missing_percent[missing_percent > 0].sort_values(ascending=False))
```

fill categorical no feature columns with "none"

```python
# fill categorical "no feature" columns with "None"
cols_none = ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 
             'BsmtExposure', 'BsmtFinType2', 'BsmtQual', 'BsmtCond', 'BsmtFinType1']


for col in cols_none:
    df[col] = df[col].fillna('none')
```

fill numeric columns with a median value

```python
# fill numeric columns with median
df['GarageYrBlt'] = df['GarageYrBlt'].fillna(df['GarageYrBlt'].median())
df['MasVnrArea'] = df['MasVnrArea'].fillna(df['MasVnrArea'].median())

# remove rows where electrical is missing
df = df.dropna(subset=['Electrical'])

print(df.isnull().sum().sum())
```

taking care of columns with alot of missing value i.e (those who had high missing percent)

```python
#drop columns that are almost entirely missing, not worth fixing
df= df.drop(columns = ['PoolQC', 'MiscFeature', 'Alley', 'Fence'])

#FireplaceQu and MasVnrType: missing likely means doesnt have one
df['FireplaceQu'] = df['FireplaceQu'].fillna('None')
df['MasVnrType'] = df['MasVnrType'].fillna('None')

# LotFrontage: numeric, and unlike garage/basement, every house HAS a lot — 
# missing here just means "wasn't recorded", so median makes sense
df['LotFrontage'] = df['LotFrontage'].fillna(df['LotFrontage'].median())

print(df.isnull().sum().sum()) #output zero means we took care of all missing values
```

saving cleaned data into a new csv

```python
df.to_csv('cleaned_house_prices.csv', index = False)
```
